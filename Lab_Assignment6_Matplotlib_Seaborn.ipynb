## Question 1

```python
import numpy as np
import matplotlib.pyplot as plt

M = float(input("Enter M: "))
x = np.linspace(-10, 10, 100)
y1 = M * x**2
y2 = M * np.sin(x)
plt.plot(x, y1, color="blue", linestyle="-", label="M*x^2")
plt.plot(x, y2, color="red", linestyle="--", label="M*sin(x)")
plt.title("Mathematical Functions")
plt.xlabel("x")
plt.ylabel("y")
plt.legend()
plt.grid(True)
plt.show()
```

## Question 2

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

subjects = ["Maths", "Physics", "Chemistry", "English", "Computer Science"]
scores = [85, 78, 92, 88, 95]
df = pd.DataFrame({"Subject": subjects, "Score": scores})
ax = sns.barplot(data=df, x="Subject", y="Score", hue="Subject", palette="Set2", legend=False)
for bar in ax.patches:
    ax.annotate(f"{bar.get_height():.0f}", (bar.get_x() + bar.get_width()/2, bar.get_height()), ha="center", va="bottom")
plt.title("Scores by Subject")
plt.xlabel("Subject")
plt.ylabel("Score")
plt.grid(axis="y")
plt.show()
```

## Question 3

```python
import numpy as np
import matplotlib.pyplot as plt

roll_number = 1024170140
np.random.seed(roll_number)
data = np.random.randn(50)
noise = np.random.randn(50)
fig, axes = plt.subplots(2, 2, figsize=(10, 7))
axes[0, 0].plot(data.cumsum())
axes[0, 0].set_title("Cumulative Sum")
axes[0, 1].scatter(range(50), data + noise)
axes[0, 1].set_title("Scatter with Noise")
axes[1, 0].plot(data)
axes[1, 0].set_title("Random Values")
axes[1, 1].hist(data)
axes[1, 1].set_title("Distribution")
for ax in axes.flat:
    ax.set_xlabel("Index")
    ax.grid(True)
plt.tight_layout()
plt.show()
```

## Question 4

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

url = "https://raw.githubusercontent.com/AnjulaMehto/MCA/main/company_sales_data.csv"
sales = pd.read_csv(url)
print(sales.head())

month_column = sales.columns[0]
profit_column = "total_profit" if "total_profit" in sales.columns else "total profit"
sns.lineplot(data=sales, x=month_column, y=profit_column, marker="o")
plt.title("Total Profit by Month")
plt.xticks(rotation=45)
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
month_column = sales.columns[0]
product_columns = [col for col in sales.columns if col != month_column and col != profit_column]
for col in product_columns:
    sns.lineplot(data=sales, x=month_column, y=col, label=col)
plt.title("Product Sales by Month")
plt.xticks(rotation=45)
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
for col in sales.columns[1:]:
    plt.figure(figsize=(8, 4))
    sns.barplot(data=sales, x=sales.columns[0], y=col)
    plt.title(col + " by Month")
    plt.xticks(rotation=45)
    plt.tight_layout()
    plt.show()
```
