## Question 3

```python
import random
random.seed(1024170140)
values = [random.randint(100, 900) for _ in range(100)]
print(values)
```

```python
odd_numbers = [value for value in values if value % 2 != 0]
even_numbers = [value for value in values if value % 2 == 0]
print("Odd count:", len(odd_numbers))
print("Even count:", len(even_numbers))
```

```python
def is_prime(n):
    if n < 2:
        return False
    for i in range(2, int(n ** 0.5) + 1):
        if n % i == 0:
            return False
    return True

primes = [value for value in values if is_prime(value)]
print("Prime count:", len(primes))
print("Prime numbers:", primes)
```

```python
from collections import Counter
counts = Counter(values)
most_common = max(counts.values())
print("Most frequent number(s):", [number for number, count in counts.items() if count == most_common])
print("Frequency:", most_common)
```

```python
digits = [int(digit) for digit in roll_number[:8]]
A = {digit * 7 for digit in digits}
B = {digit * 9 for digit in digits}
print("A:", A)
print("B:", B)
print("Union:", A.union(B))
print("Intersection:", A.intersection(B))
print("A - B:", A.difference(B))
print("B - A:", B.difference(A))
print("Symmetric difference:", A.symmetric_difference(B))
print("A subset of B:", A.issubset(B))
print("B superset of A:", B.issuperset(A))
```

```python
x = int(input("Enter a value to discard from A: "))
A.discard(x)
print(A)
```
